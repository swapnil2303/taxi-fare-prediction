```python
#STEP 10 — Calculate Trip Distance
#We have:
#pickup_latitude
#pickup_longitude
#dropoff_latitude
#dropoff_longitude

#We can calculate the approximate distance between pickup and drop-off using the Haversine formula.

from math import radians, sin, cos, sqrt, atan2

def calculate_distance(lat1, lon1, lat2, lon2):

    R = 6371  # Earth radius in KM

    lat1 = np.radians(lat1)
    lat2 = np.radians(lat2)

    dlat = lat2 - lat1
    dlon = np.radians(lon2 - lon1)

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))

    return R * c
```

```python
#Now create the distance column.
df["trip_distance"] = calculate_distance(
    df["pickup_latitude"],
    df["pickup_longitude"],
    df["dropoff_latitude"],
    df["dropoff_longitude"]
)
```

```python
#STEP 15 — Remove Unrealistic Records
#For this educational project, we can apply reasonable business rules.
df = df[
    (df["fare_amount"] > 0) &
    (df["fare_amount"] <= 200)
]
```

```python
#distace
df = df[
    (df["trip_distance"] > 0) &
    (df["trip_distance"] <= 100)
]
```

```python
#passenger count
df = df[
    (df["passenger_count"] > 0) &
    (df["passenger_count"] <= 6)
]
```

```python
df.shape
```

```python
#Important teaching point
#Don't tell students:
#"We remove outliers because they are bad."

#Instead explain:
#"We are removing values that are outside reasonable business ranges and may represent erroneous records."
```

```python
#STEP 16 — Select Features
#Remember:
#❌ Don't use fare_class.
#❌ Don't use pickup_datetime directly.
#We already extracted useful features from datetime.
#Our features can be:
#One correction from the earlier generic plan: for your actual file, fare_class should 
#be excluded from the regression features because it is a fare-related derived category and could leak information about the target.
features = [
    "passenger_count",
    "pickup_longitude",
    "pickup_latitude",
    "dropoff_longitude",
    "dropoff_latitude",
    "hour",
    "day",
    "month",
    "day_of_week",
    "year",
    "trip_distance"
]
```

```python
#Create X and y:
X = df[features]

y = df["fare_amount"]
```

```python
#check
X.head()
```

```python
#Train/Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
#STEP 18 — Linear Regression
lr_model = LinearRegression()

lr_model.fit(X_train, y_train)
```

```python
lr_pred = lr_model.predict(X_test)
```

```python
#Evaluate Linear Regression
#MAE
lr_mae = mean_absolute_error(
    y_test,
    lr_pred
)

lr_mae
```

```python
#MSE
lr_mse = mean_squared_error(
    y_test,
    lr_pred
)

lr_mse
```

```python
#RMSE
lr_rmse = np.sqrt(lr_mse)

lr_rmse
```

```python
#R²
lr_r2 = r2_score(
    y_test,
    lr_pred
)

lr_r2
```

```python
#STEP 20 — Decision Tree Regression
dt_model = DecisionTreeRegressor(
    max_depth=10,
    random_state=42
)

dt_model.fit(X_train, y_train)
```

```python
dt_pred = dt_model.predict(X_test)
```

```python
#evaluation
dt_mae = mean_absolute_error(
    y_test,
    dt_pred
)

dt_rmse = np.sqrt(
    mean_squared_error(y_test, dt_pred)
)

dt_r2 = r2_score(
    y_test,
    dt_pred
)

print("MAE :", dt_mae)
print("RMSE:", dt_rmse)
print("R2  :", dt_r2)
```

```python
#STEP 21 — Random Forest Regression
rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)
```

```python
rf_pred = rf_model.predict(X_test)
```

```python
#evaluation
rf_mae = mean_absolute_error(
    y_test,
    rf_pred
)

rf_rmse = np.sqrt(
    mean_squared_error(y_test, rf_pred)
)

rf_r2 = r2_score(
    y_test,
    rf_pred
)

print("MAE :", rf_mae)
print("RMSE:", rf_rmse)
print("R2  :", rf_r2)
```

```python
#STEP 22 — Compare Models
results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest"
    ],
    
    "MAE": [
        lr_mae,
        dt_mae,
        rf_mae
    ],
    
    "RMSE": [
        lr_rmse,
        dt_rmse,
        rf_rmse
    ],
    
    "R2": [
        lr_r2,
        dt_r2,
        rf_r2
    ]
})

results
```
